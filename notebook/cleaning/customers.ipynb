### 3.2 Data Cleaning

```python
customers["age"]=customers["age"].astype("Int64")
```

```python
customers["zip_code"] = customers["zip_code"].astype("string")
```

```python
customers["registration_date"] = pd.to_datetime(customers["registration_date"] , errors="coerce")
```

```python
customers["gender"] = customers["gender"].fillna("Unknown")
```

```python
customers["city"] = customers["city"].fillna("Unknown")
```

```python
customers["state"] = customers["state"].fillna("Unknown")
```

```python
customers["preferred_channel"] = customers["preferred_channel"].fillna("Unknown")
```

```python
text_coulmns =["gender" , "email" , "state" , "city" ,"preferred_channel" ]
for col in text_coulmns : 
    customers[col] = customers[col].str.strip().str.lower()
```

```python
customers["full_name"] = customers["full_name"].str.strip().str.title()
```
