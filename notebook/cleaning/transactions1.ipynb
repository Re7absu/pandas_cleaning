```python
transactions["product_name"] = transactions["product_name"].fillna("Unknown")
```

```python
transactions["product_category"] = transactions["product_category"].fillna("Unknown")
```

```python
transactions["store_location"]=transactions["store_location"].fillna("Unknown")
```

```python
transactions["payment_method"] = transactions["payment_method"].fillna("Unknown")
```

```python
transactions["quantity"]=transactions["quantity"].fillna(transactions["quantity"].median())
```

```python
transactions["price"] = transactions.groupby("product_name")["price"].transform(
    lambda x : x.fillna(x.median())
)
```
