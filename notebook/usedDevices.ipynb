### step 5 : Data Cleaning

```python
import ast
```

```python
ud["tags"]= ud["tags"].apply(lambda x :ast.literal_eval(x) if pd.notna else x)
```

```python

def clean_tags(tags):
    if not isinstance(tags, list):
        return tags

    tags = [
        tag.strip().lower()
        for tag in tags
        if tag.strip().lower() != "empty"
    ]
    

    tags = list(dict.fromkeys(tags))

    return tags if tags else pd.NA
```

```python
ud["tags"]=ud["tags"].apply(clean_tags)
```

```python
ud[["city" , "title"]] = ud[["city" , "title"]].apply(lambda col: col.str.strip().str.lower() )
```

```python
ud[ud["start_price"] > 1000][["start_price"]]
```

```python
ud.loc[14]
```

```python
ud.loc[14, "start_price"] = pd.NA
```

```python
df = ud.drop(columns="Unnamed: 0")
```

```python
ud.columns = (
    ud.columns
    .str.strip()
    .str.replace(r"(?<!^)(?=[A-Z])", "_", regex=True)
    .str.replace(" ", "_")
    .str.lower()
)
```
